```python
import os
import pandas as pd

BASE_DIR = '/kaggle/input/datasets/rkuo2000/uecfood256/UECFOOD256'
category_file = f'{BASE_DIR}/category.txt'

categorias = {}
with open(category_file, "r") as f:
    next(f)  # salta la línea "id   name"
    for linea in f:
        partes = linea.strip().split("\t")
        if len(partes) == 2:
            id_cat, nombre = partes
            categorias[int(id_cat)] = nombre.lower()

print(f"Total de categorías en UEC Food-256: {len(categorias)}\n")
print("═" * 50)
print("LISTA COMPLETA DE CATEGORÍAS:")
print("═" * 50)
for id_cat, nombre in sorted(categorias.items()):
    print(f"  {id_cat:3d}  {nombre}")
```
